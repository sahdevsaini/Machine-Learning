### PCA(Principle Component Analysis)

```python
PCA always try to reduce the dataset from Higher Dimension to Lower Dimwnsion

Principal component analysis, or PCA, is a dimensionality reduction method that is often used to reduce the dimensionality of large data sets, by transforming a large set of variables into a smaller one that still contains most of the information in the large set.
```

#### Benefits of PCA

- Reducing the size of data in terms of Dimension(means reducing the no. of Columns)<Br>
So that Execution of Algorithm will faster because dataset will reduce

- Visulazation(max we can see the 3d data ) <br>
    If dataset having 10 columns so it help to Visulazise the Data
